```python

import torch
from transformers import GPT2LMHeadModel, GPT2Tokenizer, set_seed

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

model_name = "gpt2"
tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name).to(device)
model.eval()

SEED = 42
```

```python
def generate_text(prompt, max_new_tokens=60, seed=SEED, **gen_kwargs):
    """Generate text from GPT-2 given a prompt and generation parameters."""
    set_seed(seed)
    input_ids = tokenizer.encode(prompt, return_tensors="pt").to(device)

    output_ids = model.generate(
        input_ids,
        max_new_tokens=max_new_tokens,
        pad_token_id=tokenizer.eos_token_id,
        **gen_kwargs
    )
    text = tokenizer.decode(output_ids[0], skip_special_tokens=True)
    return text
```

```python
opening_sentence = "The scientist opened the door to the old laboratory and"
print(opening_sentence)
```

```python
exp1 = generate_text(opening_sentence, do_sample=False)
print("EXPERIMENT 1 - Greedy decoding (baseline)")
print(exp1)
```

```python
exp2 = generate_text(opening_sentence, do_sample=True, temperature=1.5, top_k=0)
print("EXPERIMENT 2 - Temperature = 1.5")
print(exp2)
```

```python
exp3 = generate_text(opening_sentence, do_sample=True, top_k=5, temperature=1.0)
print("EXPERIMENT 3 - Top-k = 5")
print(exp3)
```

```python
exp4 = generate_text(opening_sentence, do_sample=True, top_p=0.9, top_k=0, temperature=1.0)
print("EXPERIMENT 4 - Top-p = 0.9")
print(exp4)
```

```python
exp5 = generate_text(opening_sentence, do_sample=True, repetition_penalty=2.0, top_k=50, temperature=1.0)
print("EXPERIMENT 5 - Repetition penalty = 2.0")
print(exp5)
```

```python
import pandas as pd

results_table = pd.DataFrame({
    "Experiment": [1, 2, 3, 4, 5],
    "Parameter Changed": ["Baseline (greedy)", "Temperature", "Top-k", "Top-p", "Repetition penalty"],
    "Value": ["do_sample=False", "temperature=1.5", "top_k=5", "top_p=0.9", "repetition_penalty=2.0"],
    "Generated Output": [exp1, exp2, exp3, exp4, exp5],
})
pd.set_option('display.max_colwidth', None)
results_table
```
